```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# plotting a simple function.

```

```python
# from a pandas dataframe.
batsman = pd.read_csv("/content/sharma-kohli.csv")
batsman

plt.plot(batsman["index"],batsman["V Kohli"])
```

```python
plt.plot(batsman["index"],batsman["V Kohli"])
plt.plot(batsman["index"],batsman["RG Sharma"])
```

```python
# labels title.
plt.plot(batsman["index"],batsman["V Kohli"])
plt.plot(batsman["index"],batsman["RG Sharma"])
plt.title("Rohit sharma Vs virat kohli Career Comparison")
plt.xlabel("Years")
plt.ylabel("Runs")
```

```python
# colors(hex) and line(width and style) and marker(size).
plt.plot(batsman["index"],batsman["V Kohli"],color="green")
plt.plot(batsman["index"],batsman["RG Sharma"],color = "red")
plt.title("Rohit sharma Vs virat kohli Career Comparison")
plt.xlabel("Years")
plt.ylabel("Runs")
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="solid",marker="*",markersize=10)
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dashdot")
plt.title("Rohit sharma Vs virat kohli Career Comparison")
plt.xlabel("Years")
plt.ylabel("Runs")
```

```python
plt.plot(batsman["index"],batsman["V Kohli"],linestyle="solid",marker="*",markersize=10,label="Virat")
plt.plot(batsman["index"],batsman["RG Sharma"],linestyle="dashdot",label="Sharma")
plt.title("Rohit sharma Vs virat kohli Career Comparison")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend(loc= "lower right")
```
