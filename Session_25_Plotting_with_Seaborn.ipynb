# Plots under this section
* scatterplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import plotly.express as px
```

```python
tips = sns.load_dataset("tips")
```

```python
tips
```

```python
# scatter plot --> Axis Level function
sns.scatterplot(x = "total_bill",y = "tip",data = tips)
plt.show()

```

```python
# relplot -> figure level function.
sns.relplot(data = tips , x = "total_bill", y = "tip",kind = "scatter")
```

```python
# scatter plot -> axes level function
sns.scatterplot(data = tips , x = "total_bill",y = "tip",hue="sex")
```

```python
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex",style="time")
```

```python
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex",style="time",size="size")
```

```python
# line plot
gap = px.data.gapminder()
gap
```

```python
gap = px.data.gapminder()
temp_df = gap[gap["country"] == "India"]
temp_df
```

```python
# axes level function
sns.lineplot(data = temp_df , x = "year",y = "lifeExp")
```

```python
# using relplot
sns.relplot(data=temp_df,x="year",y = "lifeExp",kind = "line")
```

```python
# hue -> style
temp_df = gap[gap["country"].isin(["India","Pakistan","China"])]
temp_df
```

```python
sns.relplot(kind = "line",data=temp_df,x="year",y="lifeExp",hue="country")
```

```python
temp_df = gap[gap["country"].isin(["India","Germany","Brazil"])]
temp_df
```

```python
sns.relplot(kind="line",data=temp_df,x="year",y="lifeExp",hue="country",style="continent")
```

```python
sns.relplot(kind="line",data=temp_df,x="year",y="lifeExp",hue="country",size="continent")
```

```python
# facet plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='sex')
```

```python
# facet plot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',row='sex')
```

```python
# facet plot -> figure level function -> work with relplot
# it will not work with scatterplot and lineplot
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='smoker',row='day')
```

```python
# col wrap
sns.relplot(data=gap, x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=3)
```

# 2. Distribution plot.
* used for univariate analysis
* used to find out the distribution
* Range of the observation
* Central Tendency
* is the data bimodal ?
* are there outliers ?

# plot under distribution plot
* histplot
* kdeplot
* rugplot


```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist',bins=20)
```

```python
# its also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables.

# countplot
sns.displot(data=tips,x="day",kind="hist")
```

```python
# hue parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
# element -> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset("titanic")
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex',element='step')
```

```python
# faceting using col and row -> not work on histplot function.
sns.displot(data=tips,x="tip",kind="hist",col="sex",element="step")
```

# kdeplot = kernal densicity estimate.
* Rather than using discrete bins, a KDE plot smooths the obersevations with a Guassian kernel,producing a continuous density estimate

```python
tips
```

```python
sns.kdeplot(data=tips,x="total_bill")
```

```python
# hue -> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill="True")
```

```python
# Rugplot
# plot marginal distribution by drawing ticks along the x and y axes.
# this function is intended to complement other plots by showing the location of individual in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate Histogram
# A bivariate histogram bins the data within rectangles that tile the plot.
# and then shows the count of observations within each rectangle with fill color.
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

# Matrix plot
* Heatmap
* clustermap

```python
gap
```

```python
# Heatmap
# plot rectangular data as a color - encoded matrix.
temp_df = gap.pivot(index='country',columns='year',values ='lifeExp')

# axes level function
sns.heatmap(temp_df)
```

```python
# annot
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot = True,linewidth = 1,cmap='viridis')
```

```python
# clustermap
# plot a matrix dataset a hierarcally-clustered heatmap.
# This function requries scipy to be available.
```

```python
iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```
